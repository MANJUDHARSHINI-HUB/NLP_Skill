```python
import pandas as pd
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from IPython.display import FileLink

df = pd.read_csv("IMDB_Dataset_CLEANED.csv")

print("=" * 60)
print("        CUSTOMER REVIEW CLEANING SYSTEM")
print("=" * 60)

print("\nDATASET INFORMATION")
print("-" * 60)
print("Number of rows    :", len(df))
print("Number of columns :", len(df.columns))

print("\nSAMPLE REVIEWS")
print("-" * 60)

negative_reviews = df[df['sentiment'] == 'negative'].head(2)
positive_reviews = df[df['sentiment'] == 'positive'].head(2)

for index, row in pd.concat([negative_reviews, positive_reviews]).iterrows():
    print("Sentiment :", row['sentiment'])
    print("Text      :", row['review'][:100], "...")
    print("-" * 60)

print("\nSENTIMENT DISTRIBUTION")
print("-" * 60)
print(df['sentiment'].value_counts())

print("\nCOLUMN NAMES")
print("-" * 60)

for column in df.columns:
    print("-", column)

print("\nMISSING VALUES")
print("-" * 60)

missing_values = df.isnull().sum()

for column in df.columns:
    print(column, ":", missing_values[column])

print("\nDUPLICATE REVIEWS")
print("-" * 60)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

df = df.drop_duplicates()

print("\nAFTER REMOVING DUPLICATES")
print("-" * 60)
print("Remaining rows:", len(df))

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_review(text):
    text = str(text)
    text = text.lower()
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'http\S+|www\S+|https\S+', ' ', text)
    text = re.sub(r'\d+', ' ', text)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    words = word_tokenize(text)
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]

    return ' '.join(words)

df['cleaned_review'] = df['review'].apply(clean_review)

print("\nREVIEW CLEANING")
print("-" * 60)
print("Review cleaning completed successfully!")

print("\nORIGINAL VS CLEANED REVIEWS")
print("=" * 60)

sample_reviews = pd.concat([
    df[df['sentiment'] == 'negative'].head(2),
    df[df['sentiment'] == 'positive'].head(2)
])

for index, row in sample_reviews.iterrows():
    print("\nSentiment:", row['sentiment'])

    print("\nOriginal:")
    print(row['review'][:200], "...")

    print("\nCleaned:")
    print(row['cleaned_review'][:200], "...")

    print("-" * 60)

print("\nFINAL DATASET INFORMATION")
print("=" * 60)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nColumns available:")

for column in df.columns:
    print("-", column)

output_file = "IMDB_Dataset_Preprocessed.csv"

df.to_csv(output_file, index=False)

print("\nEXPORT")
print("=" * 60)
print("Preprocessed dataset saved successfully!")
print("File name:", output_file)

print("\nDOWNLOAD")
print("=" * 60)

display(FileLink(output_file))

print("\n" + "=" * 60)
print("        PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)
```
